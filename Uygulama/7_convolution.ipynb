```python
import numpy as np
from PIL import Image
import matplotlib.pyplot as plt
import math
```

```python
image_path = "konak.jpg"
img = Image.open(image_path).convert("L")
img = np.array(img)
print("Görüntü Boyutları:", img.shape)
```

```python
plt.figure(figsize=(12, 8))
plt.title("Orijinal Görüntü - İzmir Konak")
plt.imshow(img, cmap="gray")
plt.show()
```

```python
filter_0 = np.array([
    [1, 0, -1],
    [1, 0, -1],
    [1, 0, -1]
])
```

```python
def apply_filter(img, filter):
    img_height, img_width = img.shape
    filter_height, filter_width = filter.shape

    # Çıktı boyutunu hesapla (Stride=1, Padding=0)
    output_height = img_height - filter_height + 1
    output_width = img_width - filter_width + 1

    filtered_img = np.zeros((output_height, output_width))

    # Filtreyi görüntü üzerinde gezdir
    for i in range(output_height):
        for j in range(output_width):
            # Görüntüden ilgili bölgeyi al
            img_subset = img[i: i+filter_height, j: j+filter_width]
            # Eleman bazında çarp ve topla
            summation = np.sum(img_subset * filter)
            filtered_img[i, j] = summation

    return filtered_img
```

```python
filtered_img_0 = apply_filter(img, filter_0)

plt.figure(figsize=(12, 8))
plt.title("Dikey Kenar Filtresi Sonucu")
plt.imshow(filtered_img_0, cmap="gray")
plt.show()
```

```python
filter_1 = np.array([
    [1, 1, 1],
    [0, 0, 0],
    [-1, -1, -1]
])
filtered_img_1 = apply_filter(img, filter_1)

plt.figure(figsize=(18, 9))
plt.subplot(1, 2, 1)
plt.title("Dikey Kenarlar")
plt.imshow(filtered_img_0, cmap="gray")
plt.subplot(1, 2, 2)
plt.title("Yatay Kenarlar")
plt.imshow(filtered_img_1, cmap="gray")
plt.tight_layout()
plt.show()
```

```python
filter_2 = np.array([
    [0.0625, 0.125, 0.0625],
    [0.125,   0.25, 0.125],
    [0.0625, 0.125, 0.0625]
])
filtered_img_2 = apply_filter(img, filter_2)

filter_3 = np.array([
    [0, -1, 0],
    [-1, 5, -1],
    [0, -1, 0]
])
filtered_img_3 = apply_filter(img, filter_3)

plt.figure(figsize=(18, 9))
plt.subplot(1, 2, 1)
plt.title("Bulanıklaştırma (Blur)")
plt.imshow(filtered_img_2, cmap="gray")
plt.subplot(1, 2, 2)
plt.title("Keskinleştirme (Sharpen)")
plt.imshow(filtered_img_3, cmap="gray")
plt.tight_layout()
plt.show()
```

```python
def convolve_with_stride(img, filter, stride=1):
    img_height, img_width = img.shape
    filter_height, filter_width = filter.shape
    
    # Stride'a göre çıktı boyutunu hesapla
    output_height = math.floor((img_height - filter_height) / stride) + 1
    output_width = math.floor((img_width - filter_width) / stride) + 1

    filtered_img = np.zeros((output_height, output_width))

    # Görüntü üzerinde stride adımlarıyla gez
    for i in range(output_height):
        for j in range(output_width):
            img_subset = img[i*stride: i*stride + filter_height,
                             j*stride: j*stride + filter_width]
            summation = np.sum(img_subset * filter)
            filtered_img[i, j] = summation

    return filtered_img
```

```python
edge_filter = np.array([
    [0, -1, 0],
    [-1, 4, -1],
    [0, -1, 0]
])

filtered_img_stride1 = convolve_with_stride(img, edge_filter, stride=1)
filtered_img_stride2 = convolve_with_stride(img, edge_filter, stride=2)

print("Stride=1 Çıktı Boyutu:", filtered_img_stride1.shape)
print("Stride=2 Çıktı Boyutu:", filtered_img_stride2.shape)

plt.figure(figsize=(18, 9))
plt.subplot(1, 2, 1)
plt.title(f"Stride = 1, Boyut: {filtered_img_stride1.shape}")
plt.imshow(filtered_img_stride1, cmap="gray")
plt.subplot(1, 2, 2)
plt.title(f"Stride = 2, Boyut: {filtered_img_stride2.shape}")
plt.imshow(filtered_img_stride2, cmap="gray")
plt.tight_layout()
plt.show()
```

```python
def max_pool(img, pool_size=2, stride=2):
    img_height, img_width = img.shape

    # Havuzlama sonrası çıktı boyutunu hesapla
    output_height = math.floor((img_height - pool_size) / stride) + 1
    output_width = math.floor((img_width - pool_size) / stride) + 1
    
    pooled_img = np.zeros((output_height, output_width))

    # Görüntü üzerinde stride adımlarıyla gez
    for i in range(output_height):
        for j in range(output_width):
            # Pencereyi al
            img_subset = img[i*stride: i*stride + pool_size, 
                             j*stride: j*stride + pool_size]
            # Penceredeki maksimum değeri al
            pooled_img[i, j] = np.max(img_subset)

    return pooled_img
```

```python
max_pooled_img = max_pool(img, pool_size=4, stride=4)

plt.figure(figsize=(18, 9))
plt.subplot(1, 2, 1)
plt.title(f"Orijinal Görüntü, Boyut: {img.shape}")
plt.imshow(img, cmap="gray")
plt.subplot(1, 2, 2)
plt.title(f"Max Pooling Sonucu, Boyut: {max_pooled_img.shape}")
plt.imshow(max_pooled_img, cmap="gray")
plt.tight_layout()
plt.show()
```

```python
def avg_pool(img, pool_size=2, stride=2):
    img_height, img_width = img.shape

    output_height = math.floor((img_height - pool_size) / stride) + 1
    output_width = math.floor((img_width - pool_size) / stride) + 1

    pooled_img = np.zeros((output_height, output_width))

    for i in range(output_height):
        for j in range(output_width):
            img_subset = img[i*stride: i*stride + pool_size, 
                             j*stride: j*stride + pool_size]
            # Penceredeki değerlerin ortalamasını al
            pooled_img[i, j] = np.average(img_subset)

    return pooled_img
```

```python
avg_pooled_img = avg_pool(img, pool_size=4, stride=4)

plt.figure(figsize=(18, 9))
plt.subplot(1, 2, 1)
plt.title(f"Max Pooling Sonucu, Boyut: {max_pooled_img.shape}")
plt.imshow(max_pooled_img, cmap="gray")
plt.subplot(1, 2, 2)
plt.title(f"Average Pooling Sonucu, Boyut: {avg_pooled_img.shape}")
plt.imshow(avg_pooled_img, cmap="gray")
plt.tight_layout()
plt.show()
```

```python
def apply_filter(img, filter):
    img_height, img_width = img.shape
    print(f"Image height: {img_height}, width: {img_width}")

    filter_height, filter_width = filter.shape
    print(f"Filter height: {filter_height}, width: {filter_width}")

    num_horizontal_movements = img_width - filter_width + 1
    num_vertical_movements = img_height - filter_height + 1

    filtered_img = np.zeros((num_vertical_movements, num_horizontal_movements))

    for i in range(num_vertical_movements):
        for j in range(num_horizontal_movements):
            img_subset = img[i: i+filter_height, j: j+filter_width]
            summation = np.sum(img_subset * filter)
            filtered_img[i, j] = summation

    return filtered_img
```

### Stride


```python
def convolve_with_stride(img, filter, stride=1):
    img_height, img_width = img.shape
    print(f"Image height: {img_height}, width: {img_width}")

    filter_height, filter_width = filter.shape
    print(f"Filter height: {filter_height}, width: {filter_width}")

    target_img_height = math.floor((img_height - filter_height) / stride) + 1
    target_img_width = math.floor((img_width - filter_width) / stride) + 1
    print(
        f"Target image height: {target_img_height}, width: {target_img_width}")

    filtered_img = np.zeros((target_img_height, target_img_width))

    for i in range(target_img_height):
        for j in range(target_img_width):
            img_subset = img[i*stride: i*stride + filter_height,
                             j*stride: j*stride+filter_width]
            summation = np.sum(img_subset * filter)
            filtered_img[i, j] = summation

    return filtered_img
```

```python
filter_4 = np.array([
    [0, -1, 0],
    [-1, 0, 1],
    [0, 1, 0]
])

filtered_img_4 = convolve_with_stride(img, filter_4, stride=1)
filtered_img_5 = convolve_with_stride(img, filter_4, stride=2)

plt.figure(figsize=(18, 18))
plt.subplot(1, 2, 1)
plt.title("Stride = 1")
plt.imshow(filtered_img_4, cmap="gray")
plt.subplot(1, 2, 2)
plt.title("Stride = 2")
plt.imshow(filtered_img_5, cmap="gray")
plt.tight_layout()
plt.show()
```

### Max Pooling


```python
def max_pool(img, pool_size=4, stride=2):
    img_height, img_width = img.shape
    print(f"Image height: {img_height}, width: {img_width}")

    target_img_height = math.floor((img_height - pool_size) / stride) + 1
    target_img_width = math.floor((img_width - pool_size) / stride) + 1
    print(
        f"Target image height: {target_img_height}, width: {target_img_width}")

    pooled_img = np.zeros((target_img_height, target_img_width))

    for i in range(target_img_height):
        for j in range(target_img_width):
            img_subset = img[i*stride: i*stride +
                             pool_size, j*stride: j*stride+pool_size]
            pooled_img[i, j] = np.max(img_subset)

    return pooled_img
```

```python
max_pooled_img = max_pool(img, pool_size=4, stride=4)

plt.figure(figsize=(18, 18))
plt.subplot(1, 2, 1)
plt.title("Original")
plt.imshow(img, cmap="gray")
plt.subplot(1, 2, 2)
plt.title("Max Pooled Image")
plt.imshow(max_pooled_img, cmap="gray")
plt.tight_layout()
plt.show()
```
